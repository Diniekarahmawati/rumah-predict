### 5. Memilih Kolom yang Relevan

Kita hanya mengambil kolom yang masuk akal memengaruhi harga. Kode di bawah dibuat "aman": kalau ada kolom yang ternyata tidak ada di dataset, kolom itu dilewati, bukan error.

```python
TARGET = "price_in_rp"   # nama kolom target (yang ingin diprediksi)

kolom_numerik_rencana = [            # daftar kolom angka yang ingin kita pakai
    "land_size_m2", "building_size_m2", "bedrooms", "bathrooms",
    "carports", "garages", "floors", "electricity",
]
kolom_kategori_rencana = [           # daftar kolom kategori/teks yang ingin kita pakai
    "city", "district", "certificate", "property_condition", "furnishing",
]

kolom_numerik = [k for k in kolom_numerik_rencana if k in df.columns]    # ambil hanya yang benar-benar ada di dataset
kolom_kategori = [k for k in kolom_kategori_rencana if k in df.columns]  # sama, untuk kolom kategori

print("Kolom numerik dipakai :", kolom_numerik)                          # tampilkan kolom numerik yang lolos
print("Kolom kategori dipakai:", kolom_kategori)                         # tampilkan kolom kategori yang lolos
print("Tidak ditemukan       :", [k for k in kolom_numerik_rencana + kolom_kategori_rencana if k not in df.columns])  # kolom rencana yang tidak ada

data = df[[TARGET] + kolom_numerik + kolom_kategori].copy()   # buat tabel baru "data" berisi target + kolom terpilih saja
data.head()                                                   # lihat 5 baris pertama tabel baru
```

#### 5.1 Memastikan kolom angka benar-benar bertipe angka

Data hasil *scraping* website sering berisi teks ikut tercampur, misalnya `"2200 mah"` di kolom listrik. Komputer tidak bisa menghitung teks, jadi kita ambil **angkanya saja**.

> ⚠️ Sel ini memakai variabel `TARGET`, `kolom_numerik`, dan `data` dari sel di bagian 5. Kalau muncul `NameError`, jalankan ulang dari atas: **Runtime → Run before** (atau `Ctrl + F8`).

```python
for kolom in [TARGET] + kolom_numerik:                     # ulangi untuk target dan setiap kolom numerik
    if not pd.api.types.is_numeric_dtype(data[kolom]):     # jika kolom BUKAN angka (berarti teks), perlu dibersihkan
        print(f"Membersihkan kolom '{kolom}' (sebelumnya teks)")   # beri tahu kolom mana yang dibersihkan
        data[kolom] = (
            data[kolom].astype(str)                        # ubah semua isi menjadi teks dulu
            .str.extract(r"(\d+\.?\d*)")[0]                # ambil bagian angka pertama saja (contoh "2200 mah" -> "2200")
        )
    data[kolom] = pd.to_numeric(data[kolom], errors="coerce")   # ubah ke angka; yang gagal diubah jadi NaN (kosong)

data[[TARGET] + kolom_numerik].dtypes                      # cek ulang: semua kolom ini sekarang harus int/float
```

### 11. Pemeriksaan Outlier (Pencilan)

Outlier = data yang nilainya jauh berbeda dari mayoritas. Kita deteksi dengan **boxplot** dan aturan **IQR**.

```python
kolom_cek = [TARGET, "land_size_m2", "building_size_m2"]          # kolom yang paling penting dicek outliernya
kolom_cek = [k for k in kolom_cek if k in data.columns]           # pakai hanya yang ada

fig, axes = plt.subplots(1, len(kolom_cek), figsize=(5 * len(kolom_cek), 5))   # satu boxplot per kolom
axes = np.array(axes).flatten()                                   # jadikan daftar

for ax, kolom in zip(axes, kolom_cek):                            # pasangkan setiap kotak grafik dengan kolomnya
    sns.boxplot(y=data[kolom], ax=ax, color="lightcoral")         # gambar boxplot vertikal
    ax.set_title(f"Boxplot {kolom}")                              # judul
    ax.set_yscale("log")                                          # skala log agar kotak tidak gepeng karena nilai ekstrem

plt.tight_layout()
plt.show()
```

```python
def hitung_outlier_iqr(seri):                     # fungsi untuk menghitung outlier dengan aturan IQR
    seri = seri.dropna()                          # buang data kosong
    q1 = seri.quantile(0.25)                      # kuartil 1: 25% data di bawah nilai ini
    q3 = seri.quantile(0.75)                      # kuartil 3: 75% data di bawah nilai ini
    iqr = q3 - q1                                 # IQR = rentang tengah 50% data
    batas_bawah = q1 - 1.5 * iqr                  # di bawah ini dianggap outlier
    batas_atas = q3 + 1.5 * iqr                   # di atas ini dianggap outlier
    jumlah = ((seri < batas_bawah) | (seri > batas_atas)).sum()   # hitung yang berada di luar batas
    return batas_bawah, batas_atas, jumlah        # kembalikan ketiga nilai

hasil = []                                        # wadah untuk menampung hasil setiap kolom
for kolom in [TARGET] + kolom_numerik:            # cek target dan semua kolom numerik
    bawah, atas, jml = hitung_outlier_iqr(data[kolom])            # hitung outlier kolom ini
    hasil.append([kolom, bawah, atas, jml, round(jml / data[kolom].notna().sum() * 100, 2)])  # simpan hasilnya

pd.DataFrame(hasil, columns=["kolom", "batas_bawah", "batas_atas", "jumlah_outlier", "persen (%)"])  # tampilkan sebagai tabel
```

**Cara membaca boxplot:** kotak = 50% data di tengah, garis dalam kotak = median, titik-titik di luar "kumis" = outlier.

**Hati-hati:** tidak semua outlier adalah kesalahan. Rumah mewah di Jakarta Selatan memang mahal, dan itu **data asli**. Yang perlu dibuang adalah yang **tidak masuk akal** (misalnya luas bangunan 1 m², harga Rp1.000).

### 14. Korelasi Antarvariabel (Heatmap)

**Korelasi** mengukur seberapa kuat dua variabel bergerak bersama, nilainya −1 sampai 1:
- mendekati **1** → sama-sama naik · mendekati **−1** → satu naik, satu turun · mendekati **0** → tidak berhubungan (secara linear)

```python
kolom_korelasi = [TARGET] + kolom_numerik                      # kolom yang dihitung korelasinya
matriks = data[kolom_korelasi].corr(method="spearman")         # hitung korelasi Spearman (lebih tahan outlier & data miring)

plt.figure(figsize=(10, 8))                                    # ukuran grafik
sns.heatmap(
    matriks,
    annot=True,                                                # tulis angka korelasi di setiap kotak
    fmt=".2f",                                                 # 2 angka di belakang koma
    cmap="coolwarm",                                           # warna: biru = negatif, merah = positif
    vmin=-1, vmax=1,                                           # rentang warna dari -1 sampai 1
    square=True,                                               # kotak berbentuk persegi
)
plt.title("Heatmap Korelasi (Spearman)")
plt.tight_layout()
plt.show()
```

```python
korelasi_ke_harga = (
    matriks[TARGET]                                            # ambil kolom korelasi terhadap harga
    .drop(TARGET)                                              # buang korelasi harga dengan dirinya sendiri (selalu 1)
    .sort_values(ascending=False)                              # urutkan dari yang paling kuat
)

korelasi_ke_harga.plot(kind="barh", color="teal")              # gambar diagram batang horizontal
plt.title("Korelasi Setiap Fitur terhadap Harga")
plt.xlabel("Korelasi Spearman")
plt.gca().invert_yaxis()                                       # yang terkuat di atas
plt.axvline(0, color="black", linewidth=0.8)                   # garis bantu di nilai 0
plt.show()

korelasi_ke_harga                                              # tampilkan juga angkanya
```

**Cara membaca:**
- Fitur dengan korelasi tertinggi terhadap harga adalah kandidat fitur paling berpengaruh.
- Perhatikan juga korelasi **antar fitur** yang sangat tinggi (misalnya kamar tidur vs kamar mandi > 0,8). Ini disebut **multikolinearitas** dan bisa mengganggu Linear Regression.

**Kenapa Spearman, bukan Pearson?** Pearson mengukur hubungan garis lurus dan sensitif terhadap outlier. Data harga kita miring dan banyak outlier, jadi Spearman (berbasis peringkat) lebih adil.

Kita pakai **batas logis** (aturan yang bisa dijelaskan dengan akal sehat), bukan rumus statistik. Batas-batas ini bisa kalian ubah, asal **bisa menjelaskan alasannya**.

Kita juga memakai **harga per m² bangunan** sebagai alat bantu deteksi, **hanya untuk filter, bukan fitur model.**

```python
# ===== BATAS LOGIS (boleh diubah, asal ada alasannya) =====
MIN_LUAS_BANGUNAN = 20          # m²; rumah tipe terkecil umumnya sekitar 21 m²
MIN_LUAS_TANAH = 15             # m²
MAX_KAMAR = 10                  # kamar tidur / kamar mandi; di atas ini kemungkinan kos-kosan atau salah input
MAX_GARASI_CARPORT = 10         # garasi / carport
MIN_HARGA = 100_000_000         # Rp100 juta
MIN_HARGA_PER_M2 = 1_000_000    # Rp1 juta per m² bangunan
MAX_HARGA_PER_M2 = 200_000_000  # Rp200 juta per m² bangunan

harga_per_m2 = data[TARGET] / data["building_size_m2"]          # harga per m² bangunan (hanya alat bantu)

aturan = {                                                      # setiap aturan menghasilkan True/False per baris
    "luas bangunan < 20 m²": data["building_size_m2"] < MIN_LUAS_BANGUNAN,
    "luas tanah < 15 m²": data["land_size_m2"] < MIN_LUAS_TANAH,
    "kamar tidur > 10": data["bedrooms"] > MAX_KAMAR,
    "kamar mandi > 10": data["bathrooms"] > MAX_KAMAR,
    "garasi > 10": data["garages"] > MAX_GARASI_CARPORT,
    "carport > 10": data["carports"] > MAX_GARASI_CARPORT,
    "harga < Rp100 juta": data[TARGET] < MIN_HARGA,
    "harga/m² < Rp1 juta": harga_per_m2 < MIN_HARGA_PER_M2,
    "harga/m² > Rp200 juta": harga_per_m2 > MAX_HARGA_PER_M2,
}

for nama, kondisi in aturan.items():                             # tampilkan berapa baris yang terkena tiap aturan
    print(f"{nama:25s}: {kondisi.sum():4d} baris")

tidak_wajar = pd.concat(aturan, axis=1).any(axis=1)              # True jika baris melanggar minimal 1 aturan
print(f"\nTotal baris tidak wajar (gabungan): {tidak_wajar.sum()}")
```

**Sebelum menghapus, LIHAT dulu datanya.** Jangan menghapus data tanpa memeriksanya.

```python
kolom_lihat = [TARGET, "building_size_m2", "land_size_m2", "bedrooms", "bathrooms", "garages", "city", "district"]  # kolom yang ditampilkan
contoh = data.loc[tidak_wajar, kolom_lihat].copy()                            # ambil baris yang tidak wajar
contoh["harga_per_m2_bangunan"] = harga_per_m2[tidak_wajar]                   # tambahkan info harga per m²
contoh.sort_values(TARGET, ascending=False)                                   # urutkan dari harga tertinggi
```

### 10. Pipeline Preprocessing: Imputasi, Log Fitur, Encoding, Scaling

Langkah sisanya kita rangkai dalam satu **Pipeline**:

| Kelompok kolom | Langkah | Alasan |
|---|---|---|
| Luas tanah, luas bangunan, listrik | isi kosong (median) → log → scaling | sangat miring seperti harga |
| Kamar, carport, garasi, lantai, lat, long | isi kosong (median) → scaling | nilai sudah dalam rentang wajar |
| Kota, sertifikat, kondisi, furnishing | isi kosong ("Tidak diketahui") → one-hot | model hanya bisa membaca angka |

**Kenapa imputasi pakai median, bukan rata-rata?** Data miring dan banyak nilai ekstrem. Median tidak terpengaruh nilai ekstrem.

**Kenapa kategori kosong diisi "Tidak diketahui", bukan modus?** Kosongnya sendiri bisa jadi informasi. Misalnya penjual tidak mengisi sertifikat, bisa jadi karena sertifikatnya bermasalah. Dengan kategori sendiri, model bisa mempelajari polanya.

**Kenapa scaling?** Linear Regression sensitif terhadap skala (luas tanah ratusan, lantai 1–3). Random Forest dan Gradient Boosting tidak terpengaruh scaling, jadi aman dipakai untuk semua model.

**Kenapa harus pakai Pipeline?** Median dan kategori **hanya dipelajari dari data training** (`fit`), lalu diterapkan ke data testing (`transform`). Kalau median dihitung dari seluruh data, informasi data testing ikut "bocor" ke training (data leakage).

```python
fitur_log = ["land_size_m2", "building_size_m2", "electricity"]            # fitur miring -> perlu log
fitur_biasa = [k for k in fitur_numerik if k not in fitur_log]             # fitur numerik lainnya

alur_log = Pipeline([                                                      # alur untuk fitur miring
    ("isi_kosong", SimpleImputer(strategy="median")),                      # 1) isi kosong dengan median
    ("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one")),  # 2) log1p = log(1 + x), aman jika x = 0
    ("skala", StandardScaler()),                                           # 3) samakan skala
])

alur_numerik = Pipeline([                                                  # alur untuk fitur numerik biasa
    ("isi_kosong", SimpleImputer(strategy="median")),                      # 1) isi kosong dengan median
    ("skala", StandardScaler()),                                           # 2) samakan skala
])

alur_kategori = Pipeline([                                                 # alur untuk fitur kategori
    ("isi_kosong", SimpleImputer(strategy="constant", fill_value="Tidak diketahui")),  # 1) kosong -> kategori sendiri
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),  # 2) ubah kategori jadi kolom 0/1 (tabel biasa)
])

preprocessor = ColumnTransformer([                                         # gabungkan ketiga alur sesuai kolomnya
    ("log", alur_log, fitur_log),
    ("numerik", alur_numerik, fitur_biasa),
    ("kategori", alur_kategori, fitur_kategori),
])

preprocessor                                                               # tampilkan diagram pipeline
```

#### 10.1 Uji coba pipeline

`fit_transform` pada **training** (belajar + terapkan), lalu `transform` saja pada **testing** (hanya terapkan).

```python
X_train_siap = preprocessor.fit_transform(X_train)          # pelajari median/kategori dari TRAINING, lalu terapkan
X_test_siap = preprocessor.transform(X_test)                # terapkan aturan yang sama ke TESTING (tanpa belajar lagi)

nama_kolom = preprocessor.get_feature_names_out()           # nama kolom hasil preprocessing
print("Bentuk X_train setelah preprocessing:", X_train_siap.shape)
print("Bentuk X_test setelah preprocessing :", X_test_siap.shape)
print("Sisa missing value di train         :", np.isnan(X_train_siap).sum())   # harus 0
print("\nKolom hasil preprocessing:")
print(list(nama_kolom))
```

- Sisa missing value harus **0**.
- Nama kolom diawali `log__`, `numerik__`, `kategori__` sesuai alur yang dilewati.

### 2. Fungsi Evaluasi

Model memprediksi **log(harga)**. Supaya metrik mudah dipahami, prediksi dikembalikan ke **Rupiah** dengan `np.exp()` sebelum dihitung.

| Metrik | Arti | Bagus jika |
|---|---|---|
| **MAE** | rata-rata selisih prediksi dengan harga asli (Rp) | kecil |
| **RMSE** | mirip MAE, tapi kesalahan besar dihukum lebih berat (Rp) | kecil |
| **R²** | seberapa besar variasi harga yang bisa dijelaskan model (0–1) | mendekati 1 |
| **MAPE** | rata-rata kesalahan dalam **persen** | kecil |

**Kenapa ada MAPE?** Error Rp500 juta itu besar untuk rumah Rp800 juta, tapi kecil untuk rumah Rp20 miliar. MAPE mengukur kesalahan secara adil untuk semua kelas harga, dan paling mudah dijelaskan ke orang awam: *"rata-rata meleset sekian persen."*

**Catatan R²:** kita hitung R² di dua skala. **R² (log)** adalah skala yang dipelajari model. **R² (Rp)** sangat dipengaruhi beberapa rumah super mahal, jadi bisa lebih rendah atau naik-turun.

```python
def evaluasi(model, X, y_log):                                    # fungsi menghitung semua metrik untuk satu model
    pred_log = model.predict(X)                                   # prediksi dalam skala log
    asli = np.exp(y_log)                                          # harga asli dalam Rupiah
    pred = np.exp(pred_log)                                       # prediksi dalam Rupiah
    return {
        "MAE (Rp)": mean_absolute_error(asli, pred),              # rata-rata selisih absolut
        "RMSE (Rp)": np.sqrt(mean_squared_error(asli, pred)),     # akar rata-rata kuadrat selisih
        "R² (Rp)": r2_score(asli, pred),                          # R² pada skala Rupiah
        "R² (log)": r2_score(y_log, pred_log),                    # R² pada skala log
        "MAPE (%)": mean_absolute_percentage_error(asli, pred) * 100,   # persentase kesalahan rata-rata
    }

def rupiah(x):                                                    # fungsi kecil untuk menampilkan angka sebagai Rupiah
    return f"Rp{x:,.0f}"

print("Fungsi evaluasi siap ✅")
```

### 4. Tiga Model & 5-Fold Cross Validation

| Model | Cara kerja singkat | Alasan dipilih |
|---|---|---|
| **Linear Regression** | mencari garis lurus terbaik: harga = a×luas + b×kamar + … | baseline yang mudah diinterpretasi; cocok karena hubungan log(luas) dengan log(harga) di EDA terlihat lurus |
| **Random Forest** | ratusan pohon keputusan dilatih pada sampel data berbeda, hasilnya dirata-rata | menangkap hubungan non-linear, tahan outlier |
| **Gradient Boosting** | pohon dibuat berurutan; setiap pohon baru memperbaiki kesalahan pohon sebelumnya | biasanya terbaik untuk data tabel; terbukti di jurnal acuan |

**5-Fold Cross Validation:** data training dibagi 5 bagian. Model dilatih 5 kali; setiap kali 1 bagian dipakai untuk menguji dan 4 bagian untuk belajar. Hasilnya lebih bisa dipercaya daripada satu kali uji, dan **data testing tetap tidak disentuh.**

```python
model_dasar = {                                                   # 3 model dengan pengaturan bawaan (default)
    "Linear Regression": LinearRegression(),
    "Random Forest": RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=-1),   # n_jobs=-1: pakai semua CPU
    "Gradient Boosting": GradientBoostingRegressor(random_state=RANDOM_STATE),
}

kfold = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)   # pembagi data menjadi 5 lipatan

hasil_cv = []                                                     # wadah hasil
for nama, model in model_dasar.items():                           # ulangi untuk setiap model
    pipa = Pipeline([("preprocessing", preprocessor), ("model", model)])   # gabungkan preprocessing + model
    skor = cross_validate(
        pipa, X_train, y_train, cv=kfold,
        scoring=["r2", "neg_mean_absolute_error"],                # metrik yang dihitung (pada skala log)
        return_train_score=True,                                  # hitung juga skor di data latih
    )
    hasil_cv.append({
        "Model": nama,
        "R² train (CV)": skor["train_r2"].mean(),                 # rata-rata R² saat belajar
        "R² validasi (CV)": skor["test_r2"].mean(),               # rata-rata R² di lipatan validasi
        "Std R² validasi": skor["test_r2"].std(),                 # seberapa stabil antar lipatan
        "MAE log validasi": -skor["test_neg_mean_absolute_error"].mean(),   # MAE di skala log
    })
    print(f"{nama} selesai")

tabel_cv = pd.DataFrame(hasil_cv).set_index("Model")
tabel_cv
```

**Cara membaca:**
- **R² validasi** paling penting: kemampuan model pada data yang belum pernah dilihat.
- **R² train jauh lebih tinggi dari R² validasi** → tanda **overfitting** (model menghafal, bukan memahami).
- **Std kecil** → performa model stabil di semua lipatan.
- **MAE log 0,20** kurang lebih berarti meleset sekitar 20%.

✍️ **Catatan kelompok:**
> *(model mana R² validasinya tertinggi? model mana yang selisih train vs validasinya paling besar?)*

### 5. Hyperparameter Tuning dengan Grid Search

**Hyperparameter** = pengaturan model yang ditentukan **sebelum** training, misalnya jumlah pohon atau kedalaman pohon. Grid Search mencoba semua kombinasi yang kita tentukan, lalu memilih yang terbaik berdasarkan 5-Fold CV.

| Model | Hyperparameter yang dicoba | Artinya |
|---|---|---|
| Random Forest | `n_estimators` | jumlah pohon |
| | `max_depth` | kedalaman maksimum pohon (`None` = tanpa batas) |
| | `min_samples_leaf` | minimal data di ujung pohon; makin besar, makin tidak mudah overfitting |
| Gradient Boosting | `n_estimators` | jumlah pohon berurutan |
| | `learning_rate` | seberapa besar setiap pohon memperbaiki kesalahan |
| | `max_depth` | kedalaman tiap pohon |
| | `subsample` | porsi data yang dipakai tiap pohon (< 1 mengurangi overfitting) |

**Linear Regression tidak di-tuning** karena tidak punya hyperparameter penting.

⏳ Sel ini butuh beberapa menit.

```python
grid_rf = {                                                       # kombinasi yang dicoba untuk Random Forest
    "model__n_estimators": [200, 400],                            # "model__" = parameter milik langkah "model" di pipeline
    "model__max_depth": [None, 20],
    "model__min_samples_leaf": [1, 3],
}
grid_gb = {                                                       # kombinasi yang dicoba untuk Gradient Boosting
    "model__n_estimators": [300, 600],
    "model__learning_rate": [0.05, 0.1],
    "model__max_depth": [3, 4],
    "model__subsample": [0.8, 1.0],
}

model_final = {}                                                  # wadah model terbaik tiap algoritma
info_tuning = {}                                                  # wadah parameter terbaik

# --- Linear Regression: langsung dilatih ---
mulai = time.time()
model_final["Linear Regression"] = Pipeline([("preprocessing", preprocessor), ("model", LinearRegression())]).fit(X_train, y_train)
info_tuning["Linear Regression"] = {"parameter": "-", "waktu (detik)": round(time.time() - mulai, 1)}

# --- Random Forest & Gradient Boosting: Grid Search ---
for nama, model, grid in [
    ("Random Forest", RandomForestRegressor(random_state=RANDOM_STATE, n_jobs=-1), grid_rf),
    ("Gradient Boosting", GradientBoostingRegressor(random_state=RANDOM_STATE), grid_gb),
]:
    mulai = time.time()
    pencari = GridSearchCV(
        Pipeline([("preprocessing", preprocessor), ("model", model)]),   # pipeline yang dicari pengaturannya
        param_grid=grid,                                          # daftar kombinasi
        cv=kfold,                                                 # 5-Fold CV yang sama
        scoring="neg_mean_absolute_error",                        # pilih kombinasi dengan MAE (log) terkecil
        n_jobs=-1,                                                # pakai semua CPU agar lebih cepat
    )
    pencari.fit(X_train, y_train)                                 # jalankan pencarian
    model_final[nama] = pencari.best_estimator_                   # simpan pipeline dengan pengaturan terbaik
    info_tuning[nama] = {
        "parameter": {k.replace("model__", ""): v for k, v in pencari.best_params_.items()},
        "waktu (detik)": round(time.time() - mulai, 1),
    }
    print(f"{nama} selesai dalam {info_tuning[nama]['waktu (detik)']} detik")
    print("   parameter terbaik:", info_tuning[nama]["parameter"])
```

✍️ **Catatan kelompok:** catat parameter terbaik setiap model. Ini masuk ke PPT bagian **H. Parameter model**.

> *(tulis catatan kalian)*

❓ **Latihan tanya jawab:** *"Kenapa Grid Search memakai data training saja, bukan data testing?"* → kalau memilih pengaturan berdasarkan data testing, model jadi "disesuaikan" dengan soal ujian, sehingga skor testing tidak jujur lagi.

### 6. Evaluasi di Data Testing & Perbandingan Model

Inilah **ujian akhir**: data testing (20%) yang sama sekali belum pernah dilihat model.

```python
baris = []                                                        # wadah hasil
for nama, model in model_final.items():                           # ulangi untuk setiap model
    tr = evaluasi(model, X_train, y_train)                        # skor di data training
    te = evaluasi(model, X_test, y_test)                          # skor di data testing
    baris.append({
        "Model": nama,
        "MAE (Rp)": te["MAE (Rp)"],
        "RMSE (Rp)": te["RMSE (Rp)"],
        "R² (Rp)": te["R² (Rp)"],
        "R² (log)": te["R² (log)"],
        "MAPE (%)": te["MAPE (%)"],
        "R² log train": tr["R² (log)"],                           # untuk cek overfitting
        "MAPE train (%)": tr["MAPE (%)"],
    })
baris.append({"Model": "Baseline (median)", **hasil_baseline})    # tambahkan baseline sebagai pembanding

perbandingan = pd.DataFrame(baris).set_index("Model")
perbandingan["Selisih R² log (train - test)"] = perbandingan["R² log train"] - perbandingan["R² (log)"]   # indikator overfitting

tampil = perbandingan.copy()                                      # salinan untuk ditampilkan rapi
for k in ["MAE (Rp)", "RMSE (Rp)"]:
    tampil[k] = tampil[k].map(rupiah)                             # format Rupiah
tampil.round(3)
```

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 4.5))                   # 3 grafik berdampingan
tiga = perbandingan.drop("Baseline (median)")                      # hanya 3 model ML
warna = ["#8da0cb", "#66c2a5", "#fc8d62"]                          # warna tiap model

ax[0].barh(tiga.index, tiga["MAPE (%)"], color=warna)              # grafik 1: MAPE
ax[0].set_title("MAPE di data testing (%)\n(makin kecil makin baik)")
ax[0].axvline(perbandingan.loc["Baseline (median)", "MAPE (%)"], color="gray", ls="--", label="baseline")
ax[0].legend()

ax[1].barh(tiga.index, tiga["MAE (Rp)"] / 1e6, color=warna)       # grafik 2: MAE dalam juta
ax[1].set_title("MAE di data testing (juta Rp)\n(makin kecil makin baik)")

y_pos = np.arange(len(tiga))                                       # grafik 3: R² train vs test
ax[2].barh(y_pos + 0.2, tiga["R² log train"], height=0.4, color="lightgray", label="train")
ax[2].barh(y_pos - 0.2, tiga["R² (log)"], height=0.4, color=warna, label="test")
ax[2].set_yticks(y_pos)
ax[2].set_yticklabels(tiga.index)
ax[2].set_xlim(0.5, 1.0)
ax[2].set_title("R² (log): train vs test\n(jarak besar = overfitting)")
ax[2].legend(loc="lower right")

for a in ax:
    a.invert_yaxis()                                               # urutan model dari atas
plt.tight_layout()
plt.show()
```

**Cara membaca:**
- Bandingkan setiap model dengan **garis putus-putus baseline**. Semua model harus jauh di bawahnya.
- **Grafik kanan:** batang abu-abu (train) jauh lebih panjang dari batang berwarna (test) → model **overfitting**.
  - Random Forest biasanya punya R² train sangat tinggi (≈0,98) karena pohonnya bisa "menghafal" data. Yang dinilai adalah **skor test**.
  - Linear Regression biasanya train dan test mirip, tapi keduanya lebih rendah → cenderung **underfitting** (terlalu sederhana).

1. **MAPE test terkecil** (kriteria utama), karena aplikasi menampilkan selisih harga dalam persen (fitur cek kewajaran).
2. **MAE test** sebagai pendukung, untuk melihat kesalahan dalam Rupiah.
3. **Selisih train–test** tidak terlalu besar, supaya model tetap andal di data baru.
4. Jika selisih performa sangat tipis (< 1 poin MAPE), pertimbangkan model yang **lebih sederhana atau lebih cepat**.

```python
kandidat = perbandingan.drop("Baseline (median)").sort_values("MAPE (%)")   # urutkan dari MAPE terkecil
print(kandidat[["MAPE (%)", "MAE (Rp)", "R² (log)", "Selisih R² log (train - test)"]].round(3).to_string())

NAMA_TERBAIK = kandidat.index[0]                                  # model dengan MAPE terkecil
model_terbaik = model_final[NAMA_TERBAIK]

selisih_mape = kandidat["MAPE (%)"].iloc[1] - kandidat["MAPE (%)"].iloc[0]   # beda dengan peringkat 2
print(f"\nModel terbaik berdasarkan MAPE: {NAMA_TERBAIK}")
print(f"Unggul {selisih_mape:.2f} poin MAPE dari {kandidat.index[1]}")
if selisih_mape < 1:
    print("⚠️ Selisihnya tipis (< 1 poin). Diskusikan: apakah tetap pilih model ini, atau model yang lebih sederhana/cepat?")
```

#### 8.2 Di Segmen Harga Mana Model Paling Sering Salah?

```python
analisis = X_test.copy()                                           # salin fitur testing
analisis["harga_asli"] = harga_test.values
analisis["harga_prediksi"] = pred_test
analisis["error_persen"] = (analisis["harga_prediksi"] - analisis["harga_asli"]) / analisis["harga_asli"] * 100   # + kemahalan, - kemurahan
analisis["error_abs_persen"] = analisis["error_persen"].abs()

segmen = pd.cut(                                                   # kelompokkan rumah berdasarkan harga asli
    analisis["harga_asli"],
    bins=[0, 1e9, 3e9, 10e9, np.inf],
    labels=["< Rp1 M", "Rp1–3 M", "Rp3–10 M", "> Rp10 M"],
)
per_segmen = analisis.groupby(segmen, observed=True).agg(
    jumlah_rumah=("harga_asli", "size"),
    MAPE_persen=("error_abs_persen", "mean"),
    rata2_error_persen=("error_persen", "mean"),                   # + berarti cenderung kemahalan
)
per_segmen
```

```python
per_kota = analisis.groupby("city").agg(                           # kesalahan per kota
    jumlah_rumah=("harga_asli", "size"),
    MAPE_persen=("error_abs_persen", "mean"),
).sort_values("MAPE_persen", ascending=False)

fig, ax = plt.subplots(1, 2, figsize=(15, 4.5))
per_segmen["MAPE_persen"].plot(kind="bar", ax=ax[0], color="#fc8d62", rot=0)
ax[0].set_title("MAPE per Segmen Harga")
ax[0].set_ylabel("MAPE (%)")
ax[0].set_xlabel("")
per_kota["MAPE_persen"].plot(kind="barh", ax=ax[1], color="#8da0cb")
ax[1].set_title("MAPE per Kota")
ax[1].set_xlabel("MAPE (%)")
ax[1].set_ylabel("")
ax[1].invert_yaxis()
plt.tight_layout()
plt.show()

per_kota
```

**Cara membaca:**
- **`rata2_error_persen` positif** → model cenderung **kemahalan** di segmen itu; **negatif** → cenderung **kemurahan**. Umumnya model "menarik ke tengah": rumah murah ditebak sedikit kemahalan, rumah sangat mahal ditebak kemurahan.
- Kota dengan data sedikit (misalnya Jakarta Pusat) biasanya punya error lebih besar karena model punya lebih sedikit contoh untuk belajar.

#### 8.4 Fitur Apa yang Paling Berpengaruh?

Kita pakai **Permutation Importance**: nilai satu fitur di data testing diacak, lalu dilihat seberapa banyak performa model turun. **Semakin besar penurunannya, semakin penting fitur itu.**

Kelebihan metode ini: hasilnya per **fitur asli** (misalnya `city`, bukan `city_Bekasi`), dan bisa dipakai untuk model apa pun.

```python
hasil_pi = permutation_importance(
    model_terbaik, X_test, y_test,                                 # model, fitur asli, target log
    n_repeats=10,                                                  # acak 10 kali per fitur lalu dirata-rata
    random_state=RANDOM_STATE,
    scoring="r2",                                                  # ukur penurunan R²
    n_jobs=-1,
)
importance = pd.Series(hasil_pi.importances_mean, index=X_test.columns).sort_values()   # urutkan

plt.figure(figsize=(9, 6))
importance.plot(kind="barh", color="#66c2a5", xerr=pd.Series(hasil_pi.importances_std, index=X_test.columns)[importance.index])
plt.title(f"Permutation Importance — {NAMA_TERBAIK}")
plt.xlabel("Penurunan R² jika fitur diacak")
plt.tight_layout()
plt.show()

importance.sort_values(ascending=False).round(4)
```

**Cara membaca:** fitur paling atas (batang terpanjang) adalah yang paling diandalkan model. Bandingkan dengan **korelasi di EDA**: apakah urutannya mirip? Biasanya luas bangunan dan lokasi (`lat`/`long`/`city`) di atas.

**Catatan:** fitur yang saling berkorelasi tinggi (misalnya luas bangunan dan kamar mandi, korelasi 0,81) bisa "berbagi" kepentingan. Kalau satu diacak, model masih bisa mengandalkan yang lain, sehingga importance keduanya terlihat lebih kecil dari seharusnya.

### 2. Memahami Cara Kerja Linear Regression

Linear Regression mencari **satu persamaan garis lurus** yang paling cocok dengan data:

`log(harga) = b0 + b1×luas_tanah + b2×luas_bangunan + b3×kamar + … + b33×kolom_33`

- `b0` = intersep (nilai dasar).
- `b1, b2, …` = **koefisien**: seberapa besar log(harga) berubah jika fitur itu naik 1 satuan (di sini 1 standar deviasi, karena fitur sudah di-*scaling*).
- Model mencari koefisien yang membuat **jumlah kuadrat kesalahan sekecil mungkin** (metode *Ordinary Least Squares*).

**Kelebihan:** sederhana, cepat, mudah dijelaskan (setiap fitur punya satu angka pengaruh).
**Kekurangan:** hanya bisa membentuk hubungan **lurus**. Pengaruh yang berbeda-beda (misalnya tambahan 50 m² di Jakarta Selatan jauh lebih mahal daripada di Bogor) tidak bisa ditangkap.

**Kenapa cocok dicoba?** Di EDA, hubungan log(luas bangunan) dan log(harga) terlihat hampir lurus. Karena luas dan harga sama-sama di-log, Linear Regression di sini sebenarnya memodelkan hubungan **persentase**, yang lebih masuk akal untuk harga rumah.

**Hyperparameter:** Linear Regression biasa tidak punya hyperparameter penting, jadi **tidak perlu Grid Search**.

### 3. Fungsi Evaluasi

Prediksi model berupa log(harga), jadi dikembalikan ke Rupiah dengan `np.exp()` sebelum dihitung. Metriknya sama persis dengan versi lain agar bisa dibandingkan.

```python
def evaluasi(model, X, y_log):                                    # hitung semua metrik untuk satu model
    pred_log = model.predict(X)
    asli, pred = np.exp(y_log), np.exp(pred_log)                  # log -> Rupiah
    return {
        "MAE (Rp)": mean_absolute_error(asli, pred),
        "RMSE (Rp)": np.sqrt(mean_squared_error(asli, pred)),
        "R² (Rp)": r2_score(asli, pred),
        "R² (log)": r2_score(y_log, pred_log),
        "MAPE (%)": mean_absolute_percentage_error(asli, pred) * 100,
    }

def rupiah(x):
    return f"Rp{x:,.0f}"

print("Fungsi evaluasi siap ✅")
```

### 4. Training + 5-Fold Cross Validation

Model dirangkai dengan preprocessor dalam satu `Pipeline`, lalu diuji dengan 5-Fold CV di data latih. Setelah itu dilatih ulang dengan seluruh data latih.

```python
model_lr = Pipeline([
    ("preprocessing", preprocessor),                              # preprocessing yang sama dengan versi lain
    ("model", LinearRegression()),                                # model versi ini
])

kfold = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)   # pembagian 5 lipatan yang sama
skor = cross_validate(model_lr, X_train, y_train, cv=kfold,
                      scoring=["r2", "neg_mean_absolute_error"], return_train_score=True)

print(f"R² latih (CV)     : {skor['train_r2'].mean():.4f}")
print(f"R² validasi (CV)  : {skor['test_r2'].mean():.4f}  ± {skor['test_r2'].std():.4f}")
print(f"MAE log validasi  : {-skor['test_neg_mean_absolute_error'].mean():.4f}")

model_lr.fit(X_train, y_train)                                    # latih final dengan semua data latih
print("\nModel Linear Regression selesai dilatih ✅")
```

✍️ **Catatan kelompok:** bandingkan R² latih dan R² validasi. Kalau keduanya mirip, model **tidak overfitting**. Tapi kalau keduanya sama-sama lebih rendah dari model lain, model cenderung **underfitting** (terlalu sederhana).

> *(tulis catatan kalian)*

### 5. Evaluasi di Data Uji

Dibandingkan juga dengan baseline (selalu menebak median).

```python
baseline = Pipeline([("preprocessing", preprocessor), ("model", DummyRegressor(strategy="median"))]).fit(X_train, y_train)

hasil = pd.DataFrame({
    "Linear Regression (uji)": evaluasi(model_lr, X_test, y_test),
    "Linear Regression (latih)": evaluasi(model_lr, X_train, y_train),
    "Baseline median (uji)": evaluasi(baseline, X_test, y_test),
}).T

tampil = hasil.copy()
for k in ["MAE (Rp)", "RMSE (Rp)"]:
    tampil[k] = tampil[k].map(rupiah)
tampil.round(3)
```

**Cara membaca:** fokus pada baris **uji**. MAPE = rata-rata prediksi meleset berapa persen. Bandingkan dengan baseline untuk membuktikan model benar-benar belajar.

### 6. Cek Overfitting / Underfitting

```python
r2_latih = hasil.loc["Linear Regression (latih)", "R² (log)"]
r2_uji = hasil.loc["Linear Regression (uji)", "R² (log)"]
mape_latih = hasil.loc["Linear Regression (latih)", "MAPE (%)"]
mape_uji = hasil.loc["Linear Regression (uji)", "MAPE (%)"]

print(f"R² (log)  latih {r2_latih:.3f}  ->  uji {r2_uji:.3f}   (selisih {r2_latih - r2_uji:.3f})")
print(f"MAPE      latih {mape_latih:.2f}% ->  uji {mape_uji:.2f}%")
if r2_latih - r2_uji < 0.03:
    print("\nSelisih kecil: model TIDAK overfitting.")
    print("Jika MAPE masih lebih tinggi dari model lain, model cenderung UNDERFITTING (terlalu sederhana).")
else:
    print("\nSelisih cukup besar: ada tanda overfitting.")
```

**Analogi:** underfitting seperti siswa yang belum paham materi: nilai latihan dan ujian sama-sama sedang. Overfitting seperti siswa yang menghafal kunci jawaban: nilai latihan tinggi, ujian turun.

### 8. Interpretasi Model

#### 8.1 Koefisien (kelebihan khas Linear Regression)

Setiap kolom punya satu koefisien. Karena target memakai log, koefisien bisa dibaca sebagai **persen perubahan harga**: `(e^koefisien − 1) × 100%` untuk kenaikan 1 standar deviasi (fitur angka) atau untuk kategori itu (fitur kategori).

```python
nama_kolom = model_lr.named_steps["preprocessing"].get_feature_names_out()     # 33 nama kolom
koef = pd.Series(model_lr.named_steps["model"].coef_, index=nama_kolom)
efek_persen = (np.exp(koef) - 1) * 100                                         # ubah ke persen perubahan harga

numerik = efek_persen[[k for k in efek_persen.index if not k.startswith("kategori__")]].sort_values()
plt.figure(figsize=(9, 5))
numerik.plot(kind="barh", color=["#d95f02" if v < 0 else "#1b9e77" for v in numerik])
plt.axvline(0, color="black", lw=0.8)
plt.title("Pengaruh fitur angka terhadap harga (per kenaikan 1 standar deviasi)")
plt.xlabel("Perubahan harga (%)"); plt.tight_layout(); plt.show()

numerik.sort_values(ascending=False).round(1)
```

**Cara membaca:** batang hijau = fitur menaikkan harga, oranye = menurunkan. Contoh: jika `log__building_size_m2` = +60%, rumah yang luas bangunannya 1 standar deviasi lebih besar diperkirakan 60% lebih mahal, *dengan fitur lain tetap*.

**Hati-hati:** karena multikolinearitas (luas bangunan, kamar mandi, dan kamar tidur saling berkorelasi tinggi), koefisien bisa terlihat aneh, misalnya kamar tidur bernilai negatif. Itu bukan berarti kamar tidur membuat rumah lebih murah: informasinya sudah diambil oleh luas bangunan.

✍️ **Catatan kelompok:**
> *(fitur angka apa yang paling menaikkan harga? adakah koefisien yang terlihat aneh?)*

```python
kota = efek_persen[[k for k in efek_persen.index if k.startswith("kategori__city_")]]       # efek tiap kota
kota.index = [k.replace("kategori__city_", "") for k in kota.index]
kota = kota.sort_values()
plt.figure(figsize=(9, 4.5))
kota.plot(kind="barh", color=["#d95f02" if v < 0 else "#1b9e77" for v in kota])
plt.axvline(0, color="black", lw=0.8)
plt.title("Efek kota terhadap harga (relatif, dengan spesifikasi sama)")
plt.xlabel("Perubahan harga (%)"); plt.tight_layout(); plt.show()

kota.sort_values(ascending=False).round(1)
```

**Cara membaca:** dengan spesifikasi rumah yang sama, kota bernilai positif cenderung lebih mahal dan yang negatif lebih murah. Angkanya relatif antar kota, bukan harga mutlak.

### 2. Memahami Cara Kerja XGBoost

XGBoost adalah **versi "lebih canggih" dari Gradient Boosting**. Idenya sama: membuat ratusan pohon keputusan kecil **secara berurutan**, dan setiap pohon baru belajar **memperbaiki kesalahan** pohon-pohon sebelumnya.

**Analogi:** seperti tim penaksir rumah yang bekerja bergiliran. Penaksir pertama memberi tebakan kasar. Penaksir kedua hanya fokus pada selisih tebakan pertama, penaksir ketiga memperbaiki sisa selisihnya, dan seterusnya. Tebakan akhir = jumlah sumbangan semua penaksir.

**Bedanya dengan Gradient Boosting biasa (sklearn):**

| Aspek | Gradient Boosting (sklearn) | XGBoost |
|---|---|---|
| Regularisasi | Tidak ada penalti khusus | Ada penalti (`reg_lambda`) agar pohon tidak terlalu rumit → mengurangi overfitting |
| Kecepatan | Lebih lambat | Lebih cepat (metode histogram, bisa paralel) |
| Data kosong (NaN) | Harus diisi dulu | Bisa menangani sendiri (di project ini tetap diisi oleh preprocessor) |
| Popularitas | Umum | Sering menang di kompetisi data tabel (Kaggle) |

**Hyperparameter penting yang dicari lewat Grid Search:**

- `n_estimators` = jumlah pohon (jumlah penaksir).
- `learning_rate` = seberapa besar tiap pohon boleh mengoreksi. Kecil = lebih hati-hati, butuh lebih banyak pohon.
- `max_depth` = kedalaman tiap pohon. Makin dalam makin rumit, makin rawan overfitting.
- `min_child_weight` = syarat minimal data di setiap cabang. Makin besar, pohon makin "berhati-hati".

Parameter tetap: `subsample=0.8` (tiap pohon hanya melihat 80% baris acak), `colsample_bytree=0.8` (tiap pohon hanya melihat 80% kolom acak), `reg_lambda=1` (penalti bawaan). Ketiganya juga membantu mencegah overfitting.

### 3. Fungsi Evaluasi

Prediksi model berupa log(harga), jadi dikembalikan ke Rupiah dengan `np.exp()` sebelum dihitung. Metriknya sama persis dengan versi lain agar bisa dibandingkan.

```python
def evaluasi(model, X, y_log):                                    # hitung semua metrik untuk satu model
    pred_log = model.predict(X)
    asli, pred = np.exp(y_log), np.exp(pred_log)                  # log -> Rupiah
    return {
        "MAE (Rp)": mean_absolute_error(asli, pred),
        "RMSE (Rp)": np.sqrt(mean_squared_error(asli, pred)),
        "R² (Rp)": r2_score(asli, pred),
        "R² (log)": r2_score(y_log, pred_log),
        "MAPE (%)": mean_absolute_percentage_error(asli, pred) * 100,
    }

def rupiah(x):
    return f"Rp{x:,.0f}"

print("Fungsi evaluasi siap ✅")
```

### 4. Grid Search + 5-Fold Cross Validation

Grid Search mencoba **semua kombinasi** hyperparameter di bawah (2 × 2 × 2 × 2 = 16 kombinasi). Tiap kombinasi diuji dengan 5-Fold CV, jadi total 80 kali training. Kombinasi dengan R² validasi tertinggi dipilih, lalu otomatis dilatih ulang dengan seluruh data latih.

⏱️ Biasanya selesai dalam **1–3 menit** di Colab.

```python
pipeline_xgb = Pipeline([
    ("preprocessing", preprocessor),                              # preprocessing yang sama dengan versi lain
    ("model", XGBRegressor(
        objective="reg:squarederror",                             # regresi biasa
        subsample=0.8, colsample_bytree=0.8, reg_lambda=1.0,      # parameter tetap (anti-overfitting)
        tree_method="hist",                                       # metode cepat
        random_state=RANDOM_STATE, n_jobs=1,
    )),
])

grid_parameter = {
    "model__n_estimators": [300, 600],          # jumlah pohon
    "model__learning_rate": [0.05, 0.1],        # besar koreksi tiap pohon
    "model__max_depth": [4, 6],                 # kedalaman pohon
    "model__min_child_weight": [1, 5],          # minimal data per cabang
}

kfold = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)   # pembagian 5 lipatan yang sama
grid = GridSearchCV(pipeline_xgb, grid_parameter, cv=kfold, scoring="r2",
                    return_train_score=True, n_jobs=-1)

mulai = time.time()
grid.fit(X_train, y_train)
print(f"Grid Search selesai dalam {time.time() - mulai:.0f} detik ✅")

model_xgb = grid.best_estimator_                                  # model terbaik (sudah dilatih ulang)
terbaik = {k.replace("model__", ""): v for k, v in grid.best_params_.items()}
print("\nKombinasi terbaik:", terbaik)
print(f"R² validasi (CV) terbaik: {grid.best_score_:.4f}")
```

```python
# Lihat 5 kombinasi teratas
hasil_grid = pd.DataFrame(grid.cv_results_)
kolom = ["param_model__n_estimators", "param_model__learning_rate", "param_model__max_depth",
         "param_model__min_child_weight", "mean_train_score", "mean_test_score", "rank_test_score"]
top5 = hasil_grid[kolom].sort_values("rank_test_score").head(5)
top5.columns = ["n_estimators", "learning_rate", "max_depth", "min_child_weight",
                "R² latih (CV)", "R² validasi (CV)", "peringkat"]
top5
```

```python
r2_latih = hasil.loc["XGBoost (latih)", "R² (log)"]
r2_uji = hasil.loc["XGBoost (uji)", "R² (log)"]
mape_latih = hasil.loc["XGBoost (latih)", "MAPE (%)"]
mape_uji = hasil.loc["XGBoost (uji)", "MAPE (%)"]

print(f"R² (log)  latih {r2_latih:.3f}  ->  uji {r2_uji:.3f}   (selisih {r2_latih - r2_uji:.3f})")
print(f"MAPE      latih {mape_latih:.2f}% ->  uji {mape_uji:.2f}%")
if r2_latih - r2_uji < 0.03:
    print("\nSelisih kecil: model TIDAK overfitting.")
elif r2_latih - r2_uji < 0.10:
    print("\nAda overfitting ringan: model sedikit 'menghafal' data latih,")
    print("tetapi skor uji masih bagus, jadi model tetap layak dipakai.")
else:
    print("\nSelisih besar: model overfitting. Coba max_depth lebih kecil atau min_child_weight lebih besar.")
```

**Analogi:** underfitting seperti siswa yang belum paham materi: nilai latihan dan ujian sama-sama sedang. Overfitting seperti siswa yang menghafal kunci jawaban: nilai latihan tinggi, ujian turun.

### 3. Menghitung Jarak

Jarak dihitung dengan **rumus Haversine**: jarak garis lurus di permukaan bumi (km) dari dua koordinat. Ini bukan jarak tempuh jalan, tapi cukup untuk membandingkan "dekat" dan "jauh".

⚠️ Karena koordinat rumah adalah titik tengah area, jaraknya **perkiraan** (bisa meleset ±1–3 km).

```python
def jarak_km(lat1, lon1, lat2, lon2):
    """Jarak Haversine (km) antara dua titik koordinat."""
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2
    return 6371 * 2 * np.arcsin(np.sqrt(a))

def titik_terdekat(tabel, kategori):
    """Untuk setiap baris tabel: nama titik acuan terdekat dari kategori tertentu + jaraknya."""
    acuan = titik_acuan[titik_acuan["kategori"] == kategori].reset_index(drop=True)
    matriks = np.column_stack([jarak_km(tabel["lat"].values, tabel["long"].values, r.lat, r.long)
                               for r in acuan.itertuples()])               # baris = rumah/area, kolom = titik acuan
    return acuan["nama"].values[matriks.argmin(axis=1)], matriks.min(axis=1)

# Contoh: jarak UI ke IPB Dramaga
print(f"Contoh: UI -> IPB Dramaga = {jarak_km(-6.3606, 106.8272, -6.5592, 106.7266):.1f} km")

for tabel in [data, area]:
    tabel["kampus_terdekat"], tabel["jarak_kampus_km"] = titik_terdekat(tabel, "Kampus")
    tabel["industri_terdekat"], tabel["jarak_industri_km"] = titik_terdekat(tabel, "Industri")

data[["city", "harga", "kampus_terdekat", "jarak_kampus_km", "industri_terdekat", "jarak_industri_km"]].head()
```

### 4. Harga Terstandarisasi per Area (bagian Machine Learning)

**Masalah:** kalau hanya memakai rata-rata harga iklan per area, perbandingannya **tidak adil**. Bisa saja area A terlihat mahal karena kebetulan rumah yang diiklankan di sana besar-besar.

**Solusi:** pakai model Gradient Boosting untuk memprediksi harga **rumah yang sama persis** (rumah standar) di setiap area. Karena spesifikasinya disamakan, perbedaan harganya murni karena **lokasi**. Cara ini mirip dengan cara menyusun indeks harga properti.

Keuntungan lain: area yang iklannya sedikit pun tetap bisa diperkirakan, karena model belajar dari area-area di sekitarnya.

```python
# Spesifikasi rumah standar (boleh diubah sesuai kebutuhan)
RUMAH_STANDAR = {
    "land_size_m2": 90, "building_size_m2": 70,      # rumah tapak sederhana 70/90
    "bedrooms": 2, "bathrooms": 1, "carports": 1, "garages": 0,
    "floors": 1, "electricity": 1300,
    "certificate": "SHM", "property_condition": "bagus", "furnishing": "unfurnished",
}

input_model = pd.DataFrame({**RUMAH_STANDAR,
                            "lat": area["lat"], "long": area["long"], "city": area["city"]})[X.columns]
area["harga_standar"] = np.exp(model.predict(input_model))                # prediksi model -> Rupiah

MIN_IKLAN = 5                                                              # batas data dianggap cukup
area["data_cukup"] = area["jumlah_iklan"] >= MIN_IKLAN

def rp(x):
    return f"Rp{x/1e9:,.2f} M" if x >= 1e9 else f"Rp{x/1e6:,.0f} jt"

print(f"Rumah standar: LT {RUMAH_STANDAR['land_size_m2']} m², LB {RUMAH_STANDAR['building_size_m2']} m², "
      f"{RUMAH_STANDAR['bedrooms']} KT, {RUMAH_STANDAR['bathrooms']} KM")
print("Median harga rumah standar di seluruh Jabodetabek:", rp(area["harga_standar"].median()))
```

Untuk setiap titik acuan, diambil semua area dalam radius **5 km**, lalu dihitung:

```python
RADIUS_KM = 5
median_jabodetabek = area["harga_standar"].median()

ringkasan = []
for r in titik_acuan.itertuples():
    jarak = jarak_km(area["lat"].values, area["long"].values, r.lat, r.long)
    sekitar = area[jarak <= RADIUS_KM]
    iklan = data[jarak_km(data["lat"].values, data["long"].values, r.lat, r.long) <= RADIUS_KM]
    if len(sekitar) == 0:
        continue
    ringkasan.append({
        "kategori": r.kategori, "nama": r.nama, "kota": r.kota,
        "jumlah_area": len(sekitar), "jumlah_iklan": len(iklan),
        "harga_standar_median": sekitar["harga_standar"].median(),
        "harga_standar_min": sekitar["harga_standar"].min(),
        "harga_standar_max": sekitar["harga_standar"].max(),
        "median_harga_iklan": iklan["harga"].median() if len(iklan) else np.nan,
        "median_harga_per_m2": iklan["harga_per_m2"].median() if len(iklan) else np.nan,
        "indeks_harga": sekitar["harga_standar"].median() / median_jabodetabek * 100,
    })
ringkasan = pd.DataFrame(ringkasan)

def label_indeks(x):
    if x >= 150: return "Premium"
    if x >= 110: return "Di atas rata-rata"
    if x >= 90:  return "Rata-rata"
    return "Terjangkau"
ringkasan["kelas_harga"] = ringkasan["indeks_harga"].map(label_indeks)

tampil = ringkasan.sort_values(["kategori", "indeks_harga"], ascending=[True, False]).copy()
for k in ["harga_standar_median", "harga_standar_min", "harga_standar_max", "median_harga_iklan"]:
    tampil[k] = tampil[k].map(lambda v: rp(v) if pd.notna(v) else "-")
tampil["median_harga_per_m2"] = tampil["median_harga_per_m2"].map(lambda v: f"Rp{v/1e6:,.1f} jt/m²" if pd.notna(v) else "-")
tampil["indeks_harga"] = tampil["indeks_harga"].round(0).astype(int)
tampil.drop(columns=["harga_standar_min", "harga_standar_max"])
```

### 8. Sistem Rekomendasi Area Strategis

Pengguna memilih **tujuan** (kampus/kawasan industri tertentu), **budget**, dan **jarak maksimal**. Sistem lalu:

1. mengambil area dalam radius jarak maksimal,
2. membuang area yang harga rumah standarnya **di atas budget**,
3. membuang area yang datanya terlalu sedikit (< 5 iklan),
4. menghitung **skor strategis (0–100)**:

`skor = 50 × (1 − jarak / jarak_maks) + 50 × (1 − harga / budget)`

Jadi area yang **dekat** dan **hemat** dapat skor tinggi. Bobot 50:50 bisa diubah, misalnya 70:30 kalau jarak lebih penting.

```python
def rekomendasi_area(tujuan, budget, jarak_maks_km=10, bobot_jarak=0.5, top_n=5):
    """Rekomendasi area untuk satu tujuan (nama kampus / kawasan industri)."""
    t = titik_acuan[titik_acuan["nama"] == tujuan]
    if t.empty:
        raise ValueError(f"'{tujuan}' tidak ada di titik_acuan. Cek ejaan namanya.")
    t = t.iloc[0]
    hasil = area.copy()
    hasil["jarak_km"] = jarak_km(hasil["lat"].values, hasil["long"].values, t.lat, t.long)
    hasil = hasil[(hasil["jarak_km"] <= jarak_maks_km) & (hasil["harga_standar"] <= budget) & hasil["data_cukup"]]
    if hasil.empty:
        print(f"Tidak ada area yang cocok. Coba naikkan budget atau jarak maksimal.")
        return hasil
    hasil["skor"] = 100 * (bobot_jarak * (1 - hasil["jarak_km"] / jarak_maks_km)
                           + (1 - bobot_jarak) * (1 - hasil["harga_standar"] / budget))
    hasil = hasil.sort_values("skor", ascending=False).head(top_n)
    return pd.DataFrame({
        "Area": hasil["area"], "Kota": hasil["city"],
        "Jarak (km)": hasil["jarak_km"].round(1),
        "Estimasi rumah 70/90": hasil["harga_standar"].map(rp),
        "Sisa budget": (budget - hasil["harga_standar"]).map(rp),
        "Jumlah iklan": hasil["jumlah_iklan"],
        "Skor strategis": hasil["skor"].round(0).astype(int),
    }).reset_index(drop=True)

print("Fungsi rekomendasi siap ✅")
```

#### Contoh 1: pekerja di Kawasan Industri Jababeka, budget Rp900 juta

```python
rekomendasi_area("Kawasan Industri Jababeka", budget=900e6, jarak_maks_km=12)
```

#### Contoh 2: dosen di Universitas Indonesia, budget Rp1,5 miliar

```python
rekomendasi_area("Universitas Indonesia", budget=1.5e9, jarak_maks_km=8)
```

#### Contoh 3: pekerja di Kawasan Industri Millennium Cikupa, budget Rp700 juta

```python
rekomendasi_area("Kawasan Industri Millennium Cikupa", budget=700e6, jarak_maks_km=12)
```

```python
import os, zipfile, sklearn, xgboost
from google.colab import files

# requirements.txt: versi harus sama dengan saat model dilatih, supaya model bisa dibuka di Streamlit
with open("requirements.txt", "w") as f:
    f.write(f"""streamlit
scikit-learn=={sklearn.__version__}
pandas
numpy
joblib
xgboost-cpu=={xgboost.__version__}
""")

file_hasil = [
    "model_terbaik.joblib", "info_model.json",
    "model_random_forest.joblib", "info_model_random_forest.json",
    "model_linear_regression.joblib", "info_model_linear_regression.json",
    "model_xgboost.joblib", "info_model_xgboost.json",
    "lokasi_district.csv", "harga_per_area.csv", "titik_acuan.csv", "ringkasan_titik_acuan.csv",
    "requirements.txt",
    "X_train.csv", "X_test.csv", "y_train.csv", "y_test.csv", "preprocessor.joblib",
]
with zipfile.ZipFile("hasil_proyek.zip", "w", zipfile.ZIP_DEFLATED) as z:
    for nama in file_hasil:
        if os.path.exists(nama):
            z.write(nama)
            print(f"✅ {nama:38s} {os.path.getsize(nama) / 1e6:6.2f} MB")
        else:
            print(f"⚠️ {nama} tidak ditemukan (bagian yang membuatnya belum dijalankan?)")

print(f"\nscikit-learn {sklearn.__version__} · xgboost {xgboost.__version__}")
files.download("hasil_proyek.zip")
```

### Selesai 🎉

Isi ZIP:
- **Untuk aplikasi (upload ke GitHub):** semua file `model_*.joblib`, `info_model*.json`, 4 file CSV lokasi/area, dan `requirements.txt`.
- **Arsip data (tidak wajib di-upload):** `X_train/X_test/y_train/y_test.csv` dan `preprocessor.joblib`.

⚠️ Kalau versi scikit-learn yang tercetak **bukan 1.6.1**, upload **semua** file model sekaligus bersama `requirements.txt` yang baru, supaya semua model dan library di Streamlit memakai versi yang sama.
